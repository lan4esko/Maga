Задание 1

```python
word = 'testtesttest'  # попробуйте также 'testing'

n = len(word)

if n % 2 == 1:
    # нечётное количество букв — выводим одну среднюю
    print(word[n // 2])
else:
    # чётное количество букв — выводим две средние
    print(word[n // 2 - 1:n // 2 + 1])
```

Задание 2

```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

# Если количество не совпадает — пары не составляем
if len(boys) != len(girls):
    print("Внимание, кто-то может остаться без пары!")
else:
    boys_sorted = sorted(boys)
    girls_sorted = sorted(girls)

    print("Идеальные пары:")
    for boy, girl in zip(boys_sorted, girls_sorted):
        print(f"{boy} и {girl}")
```
