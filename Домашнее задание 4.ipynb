## Задание 1. Преобразование дат

Нужно преобразовать даты из трёх газетных форматов в объекты `datetime`. Для этого использую `datetime.strptime()`: у каждой строки свой шаблон.


```python
from datetime import datetime

moscow_times = 'Wednesday, October 2, 2002'
guardian = 'Friday, 11.10.13'
daily_news = 'Thursday, 18 August 1977'

date_1 = datetime.strptime(moscow_times, '%A, %B %d, %Y')
date_2 = datetime.strptime(guardian, '%A, %d.%m.%y')
date_3 = datetime.strptime(daily_news, '%A, %d %B %Y')

print('The Moscow Times:', date_1)
print('The Guardian:', date_2)
print('Daily News:', date_3)
```

**Результат:** даты преобразуются в объекты `datetime`. Время в выводе равно `00:00:00`, потому что в исходных строках оно не указано.

Обозначения формата: `%A` — день недели, `%B` — название месяца, `%d` — день месяца, `%m` — номер месяца, `%Y` — четырёхзначный год, `%y` — двузначный год.


## Задание 2. Диапазон дат

Функция `date_range` должна возвращать список дат между началом и концом периода включительно. Если формат неверный или начало позже конца, результат — пустой список.


```python
from datetime import datetime, timedelta

def date_range(start_date, end_date):
    try:
        start = datetime.strptime(start_date, '%Y-%m-%d')
        end = datetime.strptime(end_date, '%Y-%m-%d')
    except ValueError:
        return []

    if start > end:
        return []

    result = []
    while start <= end:
        result.append(start.strftime('%Y-%m-%d'))
        start += timedelta(days=1)

    return result
```

### Проверка на примерах из задания


```python
print(date_range('2022-01-01', '2022-01-03'))
print(date_range('2022-01-03', '2022-01-01'))
print(date_range('2022-02-30', '2022-02-31'))
```

### Проверка результата

Сравниваю полученные значения с ожидаемыми.


```python
assert date_range('2022-01-01', '2022-01-03') == ['2022-01-01', '2022-01-02', '2022-01-03']
assert date_range('2022-01-03', '2022-01-01') == []
assert date_range('2022-02-30', '2022-02-31') == []
print('Проверки пройдены')
```
