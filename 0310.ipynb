```python
# ============================================================
#  Задание 1. Функция sum_distance
# ============================================================

def sum_distance(from_, to):
    """
    Суммирует все целые числа от from_ до to включительно.

    Параметры:
        from_ (int): начальное число
        to    (int): конечное число

    Возвращает:
        int: сумма всех чисел в диапазоне

    Пример:
        >>> sum_distance(1, 5)
        15
    """
    # Если первое число больше второго — меняем их местами
    if from_ > to:
        from_, to = to, from_

    # range(from_, to + 1) включает конечное число to
    # sum() складывает все элементы последовательности
    return sum(range(from_, to + 1))


# ---------- Проверка работы функции ----------
print("Проверка sum_distance:")
print(f"  sum_distance(1, 5)  = {sum_distance(1, 5)}")
print(f"  sum_distance(5, 1)  = {sum_distance(5, 1)}")
print(f"  sum_distance(3, 3)  = {sum_distance(3, 3)}")
print(f"  sum_distance(-2, 2) = {sum_distance(-2, 2)}")
```

```python
---

<a id="задание-2"></a>
## ✂️ Задание 2. Функция `trim_and_repeat`

**Условие:**  
Реализовать функцию, которая принимает три параметра:

- `string` — исходная строка;
- `offset` — сколько символов обрезать слева (по умолчанию `0`);
- `repetitions` — сколько раз повторить строку (по умолчанию `1`).

**Пример работы:**

| Вызов                                  | Результат                |
|----------------------------------------|--------------------------|
| `trim_and_repeat("Python")`            | `Python`                 |
| `trim_and_repeat("Python", 2)`         | `thon`                   |
| `trim_and_repeat("Python", 0, 3)`      | `PythonPythonPython`     |
| `trim_and_repeat("Python", 2, 2)`      | `thonthon`               |
```

```python
# ============================================================
#  Задание 2. Функция trim_and_repeat
# ============================================================

def trim_and_repeat(string, offset=0, repetitions=1):
    """
    Обрезает строку слева на offset символов и повторяет её repetitions раз.

    Параметры:
        string      (str): исходная строка
        offset      (int): сколько символов обрезать слева (по умолчанию 0)
        repetitions (int): сколько раз повторить строку (по умолчанию 1)

    Возвращает:
        str: полученная строка

    Пример:
        >>> trim_and_repeat("Python", 2, 2)
        'thonthon'
    """
    # string[offset:] — срез: берём всё, начиная с позиции offset
    # * repetitions — повторяем строку нужное количество раз
    return string[offset:] * repetitions


# ---------- Проверка работы функции ----------
print("Проверка trim_and_repeat:")
print(f'  trim_and_repeat("Python")          = "{trim_and_repeat("Python")}"')
print(f'  trim_and_repeat("Python", 2)       = "{trim_and_repeat("Python", 2)}"')
print(f'  trim_and_repeat("Python", 0, 3)    = "{trim_and_repeat("Python", 0, 3)}"')
print(f'  trim_and_repeat("Python", 2, 2)    = "{trim_and_repeat("Python", 2, 2)}"')
```
